```python
import pandas as pd
import numpy as np
import seaborn as sns 
import requests
import matplotlib.pyplot as plt
```

```python
url = "https://api.tfl.gov.uk/line/mode/tube/status"

r = requests.get(url)
r
```

```python
data = r.json()
```

```python
data[0]['lineStatuses'][0]['statusSeverityDescription']
```

```python
pd.DataFrame(data)
```

```python
len(data)
```

```python
rows = []

for line in data:
    name = line.get('name', "Unknown")
    statuses = line.get('lineStatuses', 'unknown')
    description = statuses[0].get('statusSeverityDescription', None)
    reason = statuses[0].get('reason', None)
    
    rows.append({
        "name": name, 
        "status": description, 
        "reason": reason,
    })
    
pd.DataFrame(rows)
```
